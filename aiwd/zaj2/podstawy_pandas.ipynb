```python
print('suma kolumny salary: ', df['salary'].sum())
```

```python
print('suma kolumny salary dla Kobiet: ', df.query("sex == 'F'")['salary'].sum())
print('suma kolumny salary dla Kobiet: ', df.query("sex == 'M'")['salary'].sum())

for lev in ['junior', 'mid', 'senior']:
    print(f'suma kolumny salary dla {lev}', {df.query(f"level == {lev!r}")["salary"].sum()})
```

```python
df_jm = df.query("level == 'junior' and sex == 'F'")['salary']
print('srednia:', df_jm.mean())
print('mediana: ', df_jm.median())
```
