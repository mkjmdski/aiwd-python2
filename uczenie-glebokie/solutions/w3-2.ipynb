```python
import tensorflow as tf
import numpy as np
import pandas as pd

from tensorflow.keras.utils import to_categorical
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, BatchNormalization, MaxPooling2D, Dropout, Flatten, Dense
from tensorflow.keras.optimizers import Adam

from sklearn.metrics import confusion_matrix

import seaborn as sns
import matplotlib.pyplot as plt
```

# Exercise 2 (CIFAR-10) - 45 minutes
The CIFAR-10 dataset consists of 60000 32x32 colour images in 10 classes, with 6000 images per class. There are 50000 training images and 10000 test images.

The dataset is divided into five training batches and one test batch, each with 10000 images. The test batch contains exactly 1000 randomly-selected images from each class. The training batches contain the remaining images in random order, but some training batches may contain more images from one class than another. Between them, the training batches contain exactly 5000 images from each class.

The classes are completely mutually exclusive. There is no overlap between automobiles and trucks. "Automobile" includes sedans, SUVs, things of that sort. "Truck" includes only big trucks. Neither includes pickup trucks.

```python
# Load the CIFAR-10 dataset
cifar10 = tf.keras.datasets.cifar10
(x_train_cifar, y_train_cifar), (x_test_cifar, y_test_cifar) = cifar10.load_data()

# Normalize the images to the range [0, 1]
x_train_cifar = x_train_cifar.astype('float32') / 255.0
x_test_cifar = x_test_cifar.astype('float32') / 255.0

# CIFAR-10 images are 32x32 color images, so the channel dimension is already 3
# No need to reshape like MNIST/Fashion MNIST, but let's confirm the shape
img_rows_cifar, img_cols_cifar, channels_cifar = 32, 32, 3
input_shape_cifar = (img_rows_cifar, img_cols_cifar, channels_cifar)

print(f'Training set dimensions for CIFAR-10: {x_train_cifar.shape}')
print(f'Test set dimensions for CIFAR-10: {x_test_cifar.shape}')

# Number of classes for CIFAR-10
num_classes_cifar = 10

# One-hot encode the labels
y_train_cifar = to_categorical(y_train_cifar, num_classes_cifar)
y_test_cifar = to_categorical(y_test_cifar, num_classes_cifar)

print(f'y_train_cifar shape: {y_train_cifar.shape}')
print(f'y_test_cifar shape: {y_test_cifar.shape}')

# Define class names for CIFAR-10
cifar_class_names = ['airplane', 'automobile', 'bird', 'cat', 'deer',
                     'dog', 'frog', 'horse', 'ship', 'truck']
```

```python
# Define the convolutional model for CIFAR-10
model_cifar = Sequential([
    Conv2D(32, (3, 3), activation='relu', kernel_initializer='he_uniform', padding='same', input_shape=input_shape_cifar),
    BatchNormalization(),
    Conv2D(32, (3, 3), activation='relu', kernel_initializer='he_uniform', padding='same'),
    BatchNormalization(),
    MaxPooling2D((2, 2)),
    Dropout(0.2),
    Conv2D(64, (3, 3), activation='relu', kernel_initializer='he_uniform', padding='same'),
    BatchNormalization(),
    Conv2D(64, (3, 3), activation='relu', kernel_initializer='he_uniform', padding='same'),
    BatchNormalization(),
    MaxPooling2D((2, 2)),
    Dropout(0.3),
    Conv2D(128, (3, 3), activation='relu', kernel_initializer='he_uniform', padding='same'),
    BatchNormalization(),
    Conv2D(128, (3, 3), activation='relu', kernel_initializer='he_uniform', padding='same'),
    BatchNormalization(),
    MaxPooling2D((2, 2)),
    Dropout(0.4),
    Flatten(),
    Dense(128, activation='relu', kernel_initializer='he_uniform'),
    BatchNormalization(),
    Dropout(0.5),
    Dense(num_classes_cifar, activation='softmax')
])

# Display the model's architecture
model_cifar.summary()

# Compile the model
model_cifar.compile(
    loss='categorical_crossentropy',
    optimizer=Adam(learning_rate=0.001),
    metrics=['accuracy']
)
```

```python
# Training the model
history_cifar = model_cifar.fit(
    x_train_cifar,
    y_train_cifar,
    epochs=25, # Increased epochs for CIFAR-10, which is more complex
    batch_size=64, # Smaller batch size for more frequent updates
    validation_split=0.15 # Using 15% for validation
)
```

```python
# Plot the training history
plt.figure(figsize=(12, 8))

# Plot training & validation accuracy values
plt.subplot(2, 1, 1)
plt.plot(history_cifar.history['accuracy'], label='Train Accuracy')
plt.plot(history_cifar.history['val_accuracy'], label='Validation Accuracy')
plt.title('CIFAR-10 Model Accuracy')
plt.ylabel('Accuracy')
plt.xlabel('Epoch')
plt.legend(loc='upper left')

# Plot training & validation loss values
plt.subplot(2, 1, 2)
plt.plot(history_cifar.history['loss'], label='Train Loss')
plt.plot(history_cifar.history['val_loss'], label='Validation Loss')
plt.title('CIFAR-10 Model Loss')
plt.ylabel('Loss')
plt.xlabel('Epoch')
plt.legend(loc='upper left')

plt.tight_layout()
plt.show()

# Evaluate the model's performance on the training data
train_loss_cifar, train_accuracy_cifar = model_cifar.evaluate(x_train_cifar, y_train_cifar, verbose=0)
print(f'CIFAR-10 Train accuracy: {train_accuracy_cifar:.4f}, Train loss: {train_loss_cifar:.4f}')

# Evaluate the model's performance on the test data
test_loss_cifar, test_accuracy_cifar = model_cifar.evaluate(x_test_cifar, y_test_cifar, verbose=0)
print(f'CIFAR-10 Test accuracy: {test_accuracy_cifar:.4f}, Test loss: {test_loss_cifar:.4f}')

# Predict probabilities on test data
predictions_cifar = model_cifar.predict(x_test_cifar)

# Convert predicted probabilities to class labels
predicted_labels_cifar = np.argmax(predictions_cifar, axis=1)

# Convert true labels from one-hot to integer labels for comparison
true_labels_cifar = np.argmax(y_test_cifar, axis=1)

# Print the first few predicted digits to verify
print("First 10 predicted labels for CIFAR-10:", predicted_labels_cifar[:10])
print("First 10 true labels for CIFAR-10:", true_labels_cifar[:10])

# Plot the first 25 images with their predicted and actual labels
plt.figure(figsize=(12, 12)) # Adjusted figure size for better visibility
for i in range(25):
    plt.subplot(5, 5, i + 1)
    plt.xticks([])
    plt.yticks([])
    plt.grid(False)

    # Get the image and true label
    img_display = x_test_cifar[i]
    true_label_idx = true_labels_cifar[i] # Original integer label
    predicted_label_idx = predicted_labels_cifar[i]

    # Set the title color based on the prediction correctness
    color_cifar = '#008800' if predicted_label_idx == true_label_idx else '#bb0000'

    # Plot the image
    plt.imshow(img_display)
    plt.title(f'{cifar_class_names[predicted_label_idx]} ({cifar_class_names[true_label_idx]})', color=color_cifar, fontsize=8) # Using class names
plt.tight_layout()
plt.show()

# Create a confusion matrix
conf_matrix_cifar = confusion_matrix(true_labels_cifar, predicted_labels_cifar)

# Plot the confusion matrix
plt.figure(figsize=(10, 8))
sns.heatmap(conf_matrix_cifar, annot=True, fmt="d", cmap="YlGnBu", cbar=False,
            xticklabels=cifar_class_names, yticklabels=cifar_class_names, linewidths=.5) # Using class names for labels
plt.title('CIFAR-10 Confusion Matrix')
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.show()
```
