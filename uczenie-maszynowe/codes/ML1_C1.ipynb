# Libs

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# NumPy

```python
sprzedaz_A = np.random.randint(80, 120, 12) * np.linspace(0.95, 1.05, 12)
sprzedaz_B = np.random.randint(60, 100, 12) * np.linspace(0.90, 1.10, 12)

print("Statystyki produktu A:",
      np.round(np.mean(sprzedaz_A), 2),
      np.round(np.median(sprzedaz_A), 2),
      np.round(np.std(sprzedaz_A), 2))
print("Statystyki produktu B:",
      np.round(np.mean(sprzedaz_B), 2),
      np.round(np.median(sprzedaz_B), 2),
      np.round(np.std(sprzedaz_B), 2))
```

# Pandas

```python
df = pd.DataFrame({'Produkt A': sprzedaz_A,
                   'Produkt B': sprzedaz_B},
                  index = pd.date_range(start = '2023-01-01', periods = 12, freq = 'M'))
print(df)
```

```python
df.iloc[5, 1] = np.nan
print(df)
```

```python
df.fillna(df.mean(),
          inplace = True)
print(df)
```

# matplotlib

```python
plt.plot(df.index, df['Produkt A'], label = 'Produkt A')
plt.plot(df.index, df['Produkt B'], label = 'Produkt B')
plt.legend()
plt.title('Sprzedaż produktów A i B')
plt.xlabel('Miesiąc')
plt.ylabel('Sprzedaż')
plt.xticks(rotation = 45)
plt.show()
```

```python
srednie = df.mean()
x_pos = np.arange(len(srednie))

plt.bar(x_pos, srednie, color = ['blue', 'orange'])
plt.xticks(x_pos, srednie.index)
plt.title('Średnia sprzedaż produktów A i B')
plt.ylabel('Średnia sprzedaż')

plt.show()
```

# Seaborn

```python
sns.kdeplot(df['Produkt A'], fill = True, label = 'Produkt A')
sns.kdeplot(df['Produkt B'], fill = True, label = 'Produkt B')
plt.legend()
plt.title('Rozkład sprzedaży produktów A i B')
plt.xlabel('')
plt.show()
```

```python
nazwy_miesiecy = ['Sty', 'Lut', 'Mar', 'Kwi', 'Maj', 'Cze', 'Lip', 'Sie', 'Wrz', 'Paź', 'Lis', 'Gru']
df.index = nazwy_miesiecy
print(df)
sns.heatmap(df, cmap = 'viridis', annot = True, fmt = '.2f', cbar = False)
plt.title('Heatmapa sprzedaży produktów A i B')
plt.yticks(rotation = 0)
plt.show()
```
